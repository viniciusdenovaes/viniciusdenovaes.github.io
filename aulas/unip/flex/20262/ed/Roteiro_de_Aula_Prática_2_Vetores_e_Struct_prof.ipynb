### Roteiro de Orientação para o Professor:
1.	Vetores (Arrays): Explicar que um vetor é uma estrutura homogênea que armazena elementos do mesmo tipo de forma contígua na memória. Enfatizar rigidamente que, em C, a indexação começa em 0 e vai até n-1. Alerte sobre o perigo de invasão de memória (ex: acessar v[10] em um vetor de tamanho 10).


Imagine que você precisa guardar as notas de 50 alunos. Criar 50 variáveis
diferentes (`nota1`, `nota2`, ..., `nota50`) seria um trabalho absurdo e
inviável de se manter. É aqui que entram os **vetores** (ou *arrays*).

Um vetor é uma estrutura de dados **homogênea** e **estática**.

·        **Homogênea:** Significa que ele só pode
armazenar elementos do **mesmo tipo**. Se você criou um vetor de inteiros,
ele só guardará inteiros; se criou de `float`,
apenas `float`.

·        **Contígua na memória:** O computador reserva
um "bloco único" e ininterrupto na memória RAM para o vetor. Os
elementos ficam colados uns nos outros, em sequência.

###Vetores em C

`int v[10];`

`int v [5] = { 5, 10, 15, 20, 25 };`

Ou

`int v[] = { 5, 10, 15, 20, 25 };``

O acesso a cada elemento do vetor é feito através de uma indexação da variável v. Observamos que, em C, a indexação de um vetor varia de zero a n-1, onde n representa a dimensão do vetor. Assim:

`v[0]	`^	acessa o primeiro elemento de v

`v[1]	`^	acessa o segundo elemento de v

`v[9]	`^	acessa o último elemento de v

Mas:

`v[10]	`^ está ERRADO (invasão de memória)



Se você declarou um vetor de tamanho $n$, a
posição do primeiro elemento é `0` e a posição do último elemento é $n-1$.

2.     **Cadeias de Caracteres (Strings):** Demonstrar
que strings em C são vetores do tipo char terminados obrigatoriamente por '\0'.
Se o delimitador for esquecido, as funções de leitura/escrita continuarão lendo
a memória indefinidamente.


* As cadeias de caracteres em C, são representadas por vetores do tipo _char_ terminadas, obrigatoriamente, pelo caractere nulo `('\0')`.

* Todas as funções que manipulam cadeias de caracteres recebem como parâmetro um vetor de _char_, processam caractere por caractere, até encontrarem o caractere nulo, que sinaliza o final da cadeia.

**3.**      **Estruturas
(Structs):**os vetores só conseguem guardar um tipo de dado (por exemplo, só números ou só caracteres). Já as *structs* permitem juntar vários tipos diferentes de dados em um único lugar, como se fosse um registro. Para usar, é importante entender como definir uma *struct* e como acessar seus dados usando o operador ponto (`.`)

* Em C, o tipo registro é conhecido como estrutura. Uma estrutura em C, serve basicamente para agrupar diversas variáveis dentro de um único contexto
* A estrutura, uma vez criada, passa a ser utilizável dentro do programa atuando como variáveis





```c
struct ponto {
  float x;
  float y;
};
int main(){
  typedef struct ponto p1,p2,p3;
  p1.x=0;
  p1.y=0;
  p2.x=1;
  p2.y=1;
  p3.x=1;
  p3.y=0;
  return 0;
}
```



ou



```c
#include <stdio.h>
typedef struct ponto {
  float x;
  float y;
}Ponto ;
int main(){
  Ponto p1,p2,p3;
  p1.x=0;
  p1.y=0;
  p2.x=1;
  p2.y=1;
  p3.x=1;
  p3.y=0;
  return 0;
}
```

3.     **Estruturas (Structs):**

````python
%%writefile exemplo.c
#include <stdio.h>

// Definição da struct
struct Produto {
    int codigo;
    float preco;
};

int main() {
    struct Produto item;

    // Atribuindo valores
    item.codigo = 101;
    item.preco = 19.99;

    // Exibindo os valores
    printf("Codigo: %d\n", item.codigo);
    printf("Preco: %.2f\n", item.preco);

    return 0;
}
````

### 4.1 Desafio Inicial: Cadastro de notas


Desenvolva um programa em C que receba a nota de 5 alunos e armazene-as em um vetor. O programa deve calcular a média aritmética da turma e, em seguida, exibir na tela apenas as notas que ficaram acima da média calculada.

````python
%%writefile desafio.c
#include <stdio.h>
int main() {
    float notas[5];
    float soma = 0.0, media = 0.0;
    int i;

    // Leitura das notas
    for(i = 0; i < 5; i++) {
        printf("Digite a nota do aluno %d: ", i + 1);
        scanf("%f", &notas[i]);
        soma += notas[i];
    }

    // Calculo da media
    media = soma / 5;
    printf("\nMedia da turma: %.2f\n", media);

    // Filtragem de notas acima da media
    printf("Notas acima da media:\n");
    for(i = 0; i < 5; i++) {
        if(notas[i] > media) {
            printf("Aluno %d: %.2f\n", i + 1, notas[i]);
        }
    }
    return 0;
}
````

````python
%%shell

gcc desafio.c -w -o programaexec

./programaexec
````

### 4.2 Ampliação do Desafio: Média da sala com alunos



Evolua o programa anterior. Agora, além da nota, precisamos armazenar a Matrícula de cada aluno. Crie uma estrutura chamada Aluno que contenha a matrícula (inteiro) e a nota (float). Instancie um vetor dessa estrutura para cadastrar 3 alunos. Calcule a média da turma e exiba a matrícula e a nota apenas dos alunos aprovados (nota maior ou igual a 6.0).

````python
%%writefile desafio.c
#include <stdio.h>
typedef struct {
    int matricula;
    float nota;
} Aluno;

int main() {
    Aluno turma[3];
    float soma = 0.0, media = 0.0;
    int i;

    // Entrada de dados
    for(i = 0; i < 3; i++) {
        printf("Digite a matricula do aluno %d: ", i + 1);
        scanf("%d", &turma[i].matricula);
        printf("Digite a nota do aluno %d: ", i + 1);
        scanf("%f", &turma[i].nota);
        soma += turma[i].nota;
    }

    media = soma / 3;
    printf("\nMedia Geral da Turma: %.2f\n", media);

    // Filtro de Alunos Aprovados (Nota >= 6.0)
    printf("\n--- Relatorio de Alunos Aprovados (Nota >= 6.0) ---\n");
    for(i = 0; i < 3; i++) {
        if(turma[i].nota >= 6.0) {
            printf("Matricula: %d | Nota: %.2f\n", turma[i].matricula, turma[i].nota);
        }
    }
    return 0;
}

````

````python
%%shell

gcc desafio.c -w -o programaexec

./programaexec
````
