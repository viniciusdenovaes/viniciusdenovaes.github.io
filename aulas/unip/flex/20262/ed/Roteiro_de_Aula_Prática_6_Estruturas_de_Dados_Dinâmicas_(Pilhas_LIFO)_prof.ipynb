## 3.2. Revisão Conceitual (20 minutos)


- Conceito de Pilha:** Estrutura linear onde as operações ocorrem apenas no topo. O último elemento a entrar é o primeiro a sair (LIFO).
  
- **Estrutura de Nós:** Explicar que a pilha é composta por nós, onde cada nó contém um dado e um ponteiro para o próximo elemento.
  
- **Operações Básicas:**
  
  - `cria()`: Inicializa a pilha com o topo apontando para `NULL`.
    
  - `push(pilha, valor)`: Insere um novo elemento no topo.
    
  - `pop(pilha)`: Remove o elemento do topo.

````python
%%writefile tadpilha.c
#include<stdlib.h>
#include<stdio.h>


typedef struct no{
  int info;
  struct no * proximo;
}No;

typedef struct pilha{
  No* topo;
} Pilha;

Pilha* cria(void){
  Pilha* p = (Pilha*)malloc(sizeof(Pilha));
  p->topo = NULL;
  return p;
}

/* Funcao de insercao no inicio: retorna a lista atualizada */
No* insere(No* lista, int num){
  No* novo_no = (No*)malloc(sizeof(No));
  novo_no->info = num;
  novo_no->proximo = lista;
  return novo_no;
}

No* retira(No* lista){
  No* aux = lista;
  lista = aux->proximo; /* retira no do inicio da lista */
  free(aux); /* libera espaco na memoria usado para alocar a variavel */
  return lista;
}

void push(Pilha* p, int v){
  p->topo = insere(p->topo, v);
}

int vazia(Pilha* p){
  return(p->topo == NULL);
}

int pop(Pilha* p){
  int v;
  if(vazia(p)){
    printf("Pilha vazia.\n");
    exit(1);
  }
  v = p->topo->info;
  p->topo = retira(p->topo);
  return v;
}

void libera(Pilha* p){
  No* q = p->topo;
  while(q != NULL){
    No* t = q->proximo;
    free(q);
    q = t;
  }
  free(p);
}

void imprime(Pilha* p){
  No* q;
  for(q = p->topo; q != NULL; q = q->proximo){
    printf("%d\n", q->info);
  }
}
void imprimep(Pilha* p){
	No* q;
	printf("\ntopo:%x ",p->topo) ;
  for(q = p->topo; q != NULL; q = q->proximo){
		printf("\n(%x)|info:%10d|prox: %x|",q, q->info,q->proximo);
	}

}
int main(void){
  Pilha* p = cria();
  push(p, 1);
  push(p, 2);
  push(p, 3);
  push(p, 4);
  push(p, 5);
  printf("Situacao original da pilha tipo lista:\n");
  imprimep(p);
  pop(p);
  pop(p);
  printf("\nSituacao atual da pilha tipo lista:\n");
  imprimep(p);
  push(p, 6);
  push(p, 7);
  push(p, 8);
  printf("\nSituacao atual da pilha tipo lista:\n");
  imprimep(p);
  pop(p);
  pop(p);
  pop(p);
  pop(p);
  printf("\nSituacao atual da pilha tipo lista:\n");
  imprimep(p);
  libera(p);
  return 0;
}
````

### 4.1 Desafio Inicial: Manipulação de Pilhas


Implemente um programa em C que:

1. Leia **N números inteiros**
2. Use:
  - **Pilha 1** → armazenar números **pares**
  - **Pilha 2** → armazenar números **ímpares**
3. Ao final, imprima:
  - Primeiro a pilha de pares
  - Depois a pilha de ímpares

````python
%%writefile desafio.c
#include <stdlib.h>
#include <stdio.h>

/* ===== TAD FORNECIDA ===== */

typedef struct no{
  int info;
  struct no * proximo;
}No;

typedef struct pilha{
  No* topo;
} Pilha;

Pilha* cria(void){
  Pilha* p = (Pilha*)malloc(sizeof(Pilha));
  p->topo = NULL;
  return p;
}

No* insere(No* lista, int num){
  No* novo_no = (No*)malloc(sizeof(No));
  novo_no->info = num;
  novo_no->proximo = lista;
  return novo_no;
}

No* retira(No* lista){
  No* aux = lista;
  lista = aux->proximo;
  free(aux);
  return lista;
}

void push(Pilha* p, int v){
  p->topo = insere(p->topo, v);
}

int vazia(Pilha* p){
  return(p->topo == NULL);
}

int pop(Pilha* p){
  int v;
  if(vazia(p)){
    printf("Pilha vazia.\n");
    exit(1);
  }
  v = p->topo->info;
  p->topo = retira(p->topo);
  return v;
}

void imprime(Pilha* p){
  No* q;
  for(q = p->topo; q != NULL; q = q->proximo){
    printf("%d\n", q->info);
  }
}

/* ===== PROGRAMA PRINCIPAL ===== */

int main(){
  Pilha *pares = cria();
  Pilha *impares = cria();

  int n, valor;

  printf("Quantidade de numeros: ");
  scanf("%d", &n);

  for(int i = 0; i < n; i++){
    printf("Digite um numero: ");
    scanf("%d", &valor);

    if(valor % 2 == 0){
      push(pares, valor);
    }else{
      push(impares, valor);
    }
  }

  printf("\nPares:\n");
  imprime(pares);

  printf("\nImpares:\n");
  imprime(impares);

  return 0;
}
````

### 4.2 Ampliação do Desafio: Ponteiros com vetores




Considere a implementação de uma **Pilha dinâmica em linguagem C**, baseada em lista encadeada, com as operações de criação, inserção (`push`), remoção (`pop`), verificação de pilha vazia e liberação de memória.

Utilizando essa estrutura de dados, desenvolva um programa que realize a conversão de um número inteiro decimal para sua representação em binário.

---

## 🔹 Requisitos

1. O programa deve ler um número inteiro decimal fornecido pelo usuário.

2. A conversão para binário deve ser realizada utilizando uma **pilha**, seguindo o seguinte procedimento:
   - Dividir sucessivamente o número por 2
   - Empilhar os restos das divisões
   - Ao final, desempilhar os valores para obter o número binário

3. Utilize as funções da pilha fornecidas:
   - `push` para inserir elementos na pilha  
   - `pop` para remover elementos da pilha  
   - `vazia` para verificar se a pilha está vazia  

4. Após o processo, o programa deve:
   - Exibir o número binário correspondente

5. Tratar o caso especial em que o número digitado seja **0**.

---

## 🔹 Exemplo de Execução

```text
Digite um numero decimal: 10
Binario: 1010

````python
%%writefile desafio.c
#include <stdlib.h>
#include <stdio.h>

typedef struct no{
  int info;
  struct no *proximo;
} No;

typedef struct pilha{
  No* topo;
} Pilha;

Pilha* cria(void){
  Pilha* p = (Pilha*)malloc(sizeof(Pilha));
  p->topo = NULL;
  return p;
}

/* Inserção (push auxiliar) */
No* insere(No* lista, int num){
  No* novo_no = (No*)malloc(sizeof(No));
  novo_no->info = num;
  novo_no->proximo = lista;
  return novo_no;
}

/* Remoção (pop auxiliar) */
No* retira(No* lista){
  No* aux = lista;
  lista = aux->proximo;
  free(aux);
  return lista;
}

void push(Pilha* p, int v){
  p->topo = insere(p->topo, v);
}

int vazia(Pilha* p){
  return (p->topo == NULL);
}

int pop(Pilha* p){
  int v;
  if(vazia(p)){
    printf("Pilha vazia.\n");
    exit(1);
  }
  v = p->topo->info;
  p->topo = retira(p->topo);
  return v;
}

void libera(Pilha* p){
  No* q = p->topo;
  while(q != NULL){
    No* t = q->proximo;
    free(q);
    q = t;
  }
  free(p);
}

/* ===== MAIN ===== */
int main(){
  int num;

  Pilha* p = cria();

  printf("Digite um numero decimal: ");
  scanf("%d", &num);

  if(num == 0){
    printf("Binario: 0\n");
    return 0;
  }

  /* Processo de conversão */
  while(num > 0){
    int resto = num % 2;
    push(p, resto);
    num = num / 2;
  }

  /* Impressão do binário */
  printf("Binario: ");
  while(!vazia(p)){
    printf("%d", pop(p));
  }

  printf("\n");

  libera(p);
  return 0;
}
````
