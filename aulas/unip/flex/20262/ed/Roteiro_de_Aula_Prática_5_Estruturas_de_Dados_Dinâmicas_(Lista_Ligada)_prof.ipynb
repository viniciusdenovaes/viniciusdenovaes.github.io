#Listas Ligadas
Estruturas Unidimensionais são um conjunto de elementos do mesmo tipo tendo uma relação linear. O relacionamento entre as informações de uma lista linear é definido somente por sua posição em relação às outras informações (EDELWEISS; GALANTE, 2009).

##Listas Ligadas Encadeadas
A organização é a lógica em que as informações podem estar separadas fisicamente, mas ligadas por referência, neste tipo de arranjo sempre existe um mínimo de um par formado entre a informação e a referência à próxima informação. Este arranjo é conhecido como lista ligada, ou lista encadeada.

A estrutura de dados mais simples é a lista linear, uma cadeia de informações, todas com a mesma estrutura, mas correlacionadas.
Cada uma dessas estruturas é chamada de nó (registro).

Na manipulação de listas, utilizam-se de três operações fundamentais:
* Inclusão;
* Busca;
* Remoção.

````python
%%writefile exemplo401.c
#include <stdio.h>
#include <stdlib.h>
typedef struct no{
  int info;
  struct no* proximo;
}No;

No* inicia() {
  return NULL;
}

No* insere(No* lista, int num){
    No* novo_no;
    int tamanho=sizeof(No);

    novo_no = (No*) malloc(tamanho);
    novo_no->info=num;
    novo_no->proximo=lista;
    return novo_no;
}
int main(){
    No* lista;
    lista=inicia();
    printf("\nendereço alocado para lista %x ",lista);
    lista=insere(lista,23);
    printf("\nendereço alocado para lista %x conteudo %d proximo:%x",
                                    lista,lista->info,lista->proximo);

    lista=insere(lista,45);
    printf("\nendereço alocado para lista %x conteudo %d proximo:%x",
                                    lista,lista->info,lista->proximo);

    lista=insere(lista,56);
    printf("\nendereço alocado para lista %x conteudo %d proximo:%x",
                                    lista,lista->info,lista->proximo);

    lista=insere(lista,95);
    printf("\nendereço alocado para lista %x conteudo %d proximo:%x",
                                    lista,lista->info,lista->proximo);

    return 0;
}
````

### 4.1 Desafio Inicial: Montagem do TAD e manipulação de Listas


A partir dos exemplos práticos, criar uma TAD (Tipo Abstrato de Dados), com as seguintes funções:

*   No* inicia()
*   No* insere(No* lista, int num)
*   No* busca(No* lista,int num)
*   void imprime(No* lista)
*   No* retira(No* lista, int num)


Utilize o seguinte estrutura de Nó:
```
typedef struct no{
  int info;
  struct no* proximo;
}No;
```



Monte na função main() duas listas (lista1 e lista2) e faça as seguintes operações:
* cria e inicia lista1;
* insere na lista1 1;
* insere na lista1 2;
* insere na lista1 3;
* insere na lista1 4;
* imprime a lista1;
* cria e inicia lista2;
* insere na lista2 10;
* insere na lista2 20;
* insere na lista1 5;
* insere na lista2 30;
* remove da lista1 4;
* insere na lista2 40;
* imprime a lista1;
* imprime a lista2;


````python
%%writefile desafio.c
#include <stdio.h>
#include <stdlib.h>

// Definição da estrutura do Nó
typedef struct no {
    int info;
    struct no* proximo;
} No;

// Função de inicialização
No* inicia() {
    return NULL;
}

// Função de inserção (no início da lista)
No* insere(No* lista, int num) {
    No* novo_no;
    int tamanho = sizeof(No);

    novo_no = (No*) malloc(tamanho);
    novo_no->info = num;
    novo_no->proximo = lista;
    return novo_no;
}

// Função de busca
No* busca(No* lista, int num) {
    No* aux;
    for (aux = lista; aux != NULL; aux = aux->proximo) {
        if (aux->info == num)
            return aux;
    }
    return NULL;
}

// Função de impressão
void imprime(No* lista) {
    No* aux;
    if (lista == NULL) {
        printf("Lista vazia.\n");
        return;
    }
    for (aux = lista; aux != NULL; aux = aux->proximo) {
        printf("\nEndereco alocado: %x | Conteudo: %d | Proximo: %x",
               aux, aux->info, aux->proximo);
    }
    printf("\n");
}

// Função de remoção
No* retira(No* lista, int num) {
    No* anterior = NULL; /* ponteiro para o no anterior */
    No* atual = lista;   /* ponteiro auxiliar para percorrer lista */

    /* procura o no na lista, guardando o no anterior*/
    for (atual = lista; atual != NULL; atual = atual->proximo) {
        if (atual->info == num) {
            break;  // achou
        } else {
            anterior = atual;
        }
    }

    /* Verifica se achou o no */
    if (atual == NULL) {
        return lista; /* nao achou, retorna lista original */
    }

    /* achou, retira o no da lista */
    if (anterior == NULL) {
        lista = atual->proximo; /* retira no do inicio da lista */
    } else {
        anterior->proximo = atual->proximo; /* retira no do meio/fim da lista */
    }

    free(atual); /* libera espaco na memoria */
    return lista;
}


int main() {
    // cria e inicia lista1;
    No* lista1 = inicia();

    // insere na lista1 1;
    lista1 = insere(lista1, 1);
    // insere na lista1 2;
    lista1 = insere(lista1, 2);
    // insere na lista1 3;
    lista1 = insere(lista1, 3);
    // insere na lista1 4;
    lista1 = insere(lista1, 4);

    // imprime a lista1;
    printf("--- IMPRESSAO: Lista 1 (Apos insercoes iniciais) ---");
    imprime(lista1);

    // cria e inicia lista2;
    No* lista2 = inicia();

    // insere na lista2 10;
    lista2 = insere(lista2, 10);
    // insere na lista2 20;
    lista2 = insere(lista2, 20);
    // insere na lista1 5;
    lista1 = insere(lista1, 5);
    // insere na lista2 30;
    lista2 = insere(lista2, 30);

    // remove da lista1 4;
    lista1 = retira(lista1, 4);

    // insere na lista2 40;
    lista2 = insere(lista2, 40);

    // imprime a lista1;
    printf("\n--- IMPRESSAO: Lista 1 (Final) ---");
    imprime(lista1);

    // imprime a lista2;
    printf("\n--- IMPRESSAO: Lista 2 (Final) ---");
    imprime(lista2);

    // Liberar a memória antes de fechar o programa (Boa prática)
    while(lista1 != NULL) lista1 = retira(lista1, lista1->info);
    while(lista2 != NULL) lista2 = retira(lista2, lista2->info);

    return 0;
}
````

````python
%%shell

gcc desafio.c -w -o programa

./programa
````

````python
%%writefile desafio.c
#include <stdio.h>
#include <stdlib.h>

// Definição da estrutura Aluno
typedef struct aluno {
    int matricula;
    float nota;
} Aluno;

// Definição da estrutura do Nó (substituindo 'int info' por 'Aluno info')
typedef struct no {
    Aluno info;
    struct no* proximo;
} No;

// Função de inicialização
No* inicia() {
    return NULL;
}

// Função de inserção (no início da lista)
No* insere(No* lista, Aluno dados_aluno) {
    No* novo_no;
    int tamanho = sizeof(No);

    novo_no = (No*) malloc(tamanho);
    novo_no->info = dados_aluno; // Copia a estrutura Aluno inteira
    novo_no->proximo = lista;
    return novo_no;
}

// Função de busca alterada para localizar pela MATRÍCULA
No* busca(No* lista, int matricula) {
    No* aux;
    for (aux = lista; aux != NULL; aux = aux->proximo) {
        if (aux->info.matricula == matricula)
            return aux; // Retorna o nó onde o aluno foi encontrado
    }
    return NULL;
}

// Função de impressão adaptada para os dados do Aluno
void imprime(No* lista) {
    No* aux;
    if (lista == NULL) {
        printf("Lista vazia.\n");
        return;
    }
    for (aux = lista; aux != NULL; aux = aux->proximo) {
        printf("\nEndereco alocado: %p | Matricula: %d | Nota: %.1f | Proximo: %p",
               (void*)aux, aux->info.matricula, aux->info.nota, (void*)aux->proximo);
    }
    printf("\n");
}

// Função de remoção alterada para localizar pela MATRÍCULA
No* retira(No* lista, int matricula) {
    No* anterior = NULL; /* ponteiro para o no anterior */
    No* atual = lista;   /* ponteiro auxiliar para percorrer lista */

    /* procura o no na lista, guardando o no anterior*/
    for (atual = lista; atual != NULL; atual = atual->proximo) {
        if (atual->info.matricula == matricula) {
            break;  // achou
        } else {
            anterior = atual;
        }
    }

    /* Verifica se achou o no */
    if (atual == NULL) {
        return lista; /* nao achou, retorna lista original */
    }

    /* achou, retira o no da lista */
    if (anterior == NULL) {
        lista = atual->proximo; /* retira no do inicio da lista */
    } else {
        anterior->proximo = atual->proximo; /* retira no do meio/fim da lista */
    }

    free(atual); /* libera espaco na memoria */
    return lista;
}

// --- FUNÇÃO MAIN ---
int main() {
    // No main crie a lista salaLista
    No* salaLista = inicia();

    // Variável auxiliar para criar os alunos antes de inserir
    Aluno alu;

    // insira o aluno matricula 1 nota 5.5;
    alu.matricula = 1; alu.nota = 5.5;
    salaLista = insere(salaLista, alu);

    // insira o aluno matricula 2 nota 7.5;
    alu.matricula = 2; alu.nota = 7.5;
    salaLista = insere(salaLista, alu);

    // insira o aluno matricula 3 nota 10.0;
    alu.matricula = 3; alu.nota = 10.0;
    salaLista = insere(salaLista, alu);

    // insira o aluno matricula 4 nota 1.0;
    alu.matricula = 4; alu.nota = 1.0;
    salaLista = insere(salaLista, alu);

    // insira o aluno matricula 5 nota 6.0;
    alu.matricula = 5; alu.nota = 6.0;
    salaLista = insere(salaLista, alu);

    // Busque o aluno matricula 3 e mostre a sua nota;
    printf("--- BUSCANDO ALUNO 3 ---\n");
    No* aluno_encontrado = busca(salaLista, 3);
    if (aluno_encontrado != NULL) {
        printf("Aluno de matricula %d encontrado! Nota: %.1f\n",
               aluno_encontrado->info.matricula, aluno_encontrado->info.nota);
    } else {
        printf("Aluno nao encontrado.\n");
    }

    // remova o aluno 3
    printf("\n--- REMOVENDO ALUNO 3 ---\n");
    salaLista = retira(salaLista, 3);

    // liste a sala
    printf("\n--- LISTAGEM FINAL DA SALA ---");
    imprime(salaLista);

    // Limpeza da memória ao encerrar
    while(salaLista != NULL) {
        salaLista = retira(salaLista, salaLista->info.matricula);
    }

    return 0;
}
````
