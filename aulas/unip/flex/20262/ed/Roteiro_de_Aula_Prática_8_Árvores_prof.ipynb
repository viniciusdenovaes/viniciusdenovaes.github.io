#Árvore



* As árvores são estruturas de dados multidimensionais que permite a representação de hierarquias ou a representação em vários níveis.
* Uma árvore é composta por um conjunto de nós, que podem, ou não, ter ramificações.
* Existe um nó denominado raiz, que pode ramificar (ou não) em subárvores, cujas raízes são ligadas, diretamente, a essa raiz e, assim, sucessivamente.


#Árvores Binárias



Uma árvore binária é um caso especial de árvore em que um pai tem, no
máximo, dois filhos (zero, um ou dois).

* Uma árvore binária é um caso especial de árvore em que um pai tem, no máximo, dois filhos (zero, um ou dois).
* Em uma árvore binária, o caminho percorrido para sair de um nó e chegar a outro sempre será único.
* A propriedade fundamental da árvore binária é que, da raiz até qualquer um dos seus nós, haverá somente um caminho.
* A quantidade de nós percorridos da raiz (sem contá-la) até a folha mais distante determina a altura da árvore.
* Uma árvore apenas com a raiz tem altura zero

De maneira recursiva, define-se uma árvore binária como sendo:
* Uma árvore vazia;
* Um nó raiz com uma das subárvores, uma identificada como a subárvore da direita (sad) ou uma identificada como a subárvore da esquerda (sae);
* Um nó raiz tendo duas subárvore, identificadas como a subárvore da direita (sad) e a subárvore da esquerda (sae).

````python
%%writefile exemplo801.c
#include <stdio.h>
#include<stdlib.h>

struct arv {
	int info;
	struct arv* esq;
	struct arv* dir;
};
typedef struct arv Arv;


int vazia(Arv* a)
{
	return a == NULL;
}
Arv* inicializa(void)
{
	return NULL;
}

Arv* cria(int c, Arv* esq, Arv* dir){
	Arv* p = (Arv*)malloc(sizeof(Arv));
	p->info = c;
	p->esq = esq;
	p->dir = dir;
	return p;
}

void imprime(Arv* a)
{
	if (!vazia(a)){
		printf("%d ", a->info); /* mostra raiz */
		imprime(a->esq); /* mostra esq */
		imprime(a->dir); /* mostra dir */
	}
}
void imprimec(Arv* a)/* impressao na forma <  a <s>,<n> > em pre ordem*/
{
	printf("<");
	if (!vazia(a)){
		printf("%d ", a->info); /* mostra raiz */
		imprimec(a->esq); /* mostra esq */
		printf(",");
		imprimec(a->dir); /* mostra dir */

	}
	printf(">");
}
void imprimepo(Arv* a)/* impressao na forma Pos Ordem>*/
{
	if (!vazia(a)){
		imprimepo(a->esq); /* mostra esq */
		imprimepo(a->dir); /* mostra dir */
		printf("%d ", a->info); /* mostra raiz */
	}
}

void imprimeos(Arv* a)/* impressao na forma Ordem simetrica*/
{
	if (!vazia(a)){
		imprimeos(a->esq); /* mostra esq */
		printf("%d ", a->info); /* mostra raiz */
		imprimeos(a->dir); /* mostra dir */
	}
}

Arv* libera(Arv* a){
	if (!vazia(a)){
		libera(a->esq); /* libera esq */
		libera(a->dir); /* libera dir */
		free(a); /* libera raiz */
	}
	return NULL;
}

int busca(Arv* a, int c){
	if (vazia(a))
		return 0; /* árvore vazia: não encontrou */
	else
		return a->info == c || busca(a->esq, c) || busca(a->dir, c);
}


Arv* insere(Arv* a, int v)
{
	if (a == NULL)
		a = cria(v, NULL, NULL);
	else
	if (v < a->info)
		a->esq = insere(a->esq, v);
	else
		a->dir = insere(a->dir, v);
	return a;
}

Arv* retira(Arv* r, int v)
{
	if (r == NULL)
		return NULL;
	else{
		if (r->info > v)
			r->esq = retira(r->esq, v);
		else{
			if (r->info < v)
				r->dir = retira(r->dir, v);
			else {	/* achou o elemento */
				if (r->esq == NULL && r->dir == NULL) { /* elemento sem filhos */
					free(r);
					r = NULL;
				}
				else{
					if (r->esq == NULL) {	/* só tem filho à direita */
						Arv* t = r;
						r = r->dir;
						free(t);
					}
					else{
						if (r->dir == NULL) {	/* só tem filho à esquerda */
							Arv* t = r;
							r = r->esq;
							free(t);
						}
						else {	/* tem os dois filhos */
							Arv* pai = r;
							Arv* f = r->esq;
							while (f->dir != NULL) {
								pai = f;
								f = f->dir;
							}
							r->info = f->info;	/* troca as informações */
							f->info = v;
							r->esq = retira(r->esq, v);
						}
					}
				}
			}
		}
	}
	return r;
}


int main(){
	Arv* ABB = insere(inicializa(), 50);
	insere(ABB, 25);
	insere(ABB, 75);
	insere(ABB, 12);
	insere(ABB, 60);

	imprimeos(ABB);

	insere(ABB, 63);
	insere(ABB, 17);
	insere(ABB, 7);
	insere(ABB, 90);
	insere(ABB, 9);
	insere(ABB, 10);
	insere(ABB, 11);
	insere(ABB, 3);
	insere(ABB, 1);
	insere(ABB, 2);
	insere(ABB, 5);
	insere(ABB, 4);
	insere(ABB, 6);
	insere(ABB, 8);
	insere(ABB, 68);


	imprimeos(ABB);

	retira(ABB, 90);

	imprimeos(ABB);

	retira(ABB, 2);

	imprimeos(ABB);

	retira(ABB, 12);

	imprimeos(ABB);
	return 0;
}
````

````python
%%writefile desafio.c
#include <stdio.h>
#include <stdlib.h>

// ==========================================
// 1. DEFINIÇÃO DA TAD ORIGINAL
// ==========================================

typedef struct arv {
   int info;
   struct arv* esq;
   struct arv* dir;
} Arv;

int vazia(Arv* a) {
  return a == NULL;
}

Arv* inicia() {
   return NULL;
}

Arv* cria(int c, Arv* sae, Arv* sad) {
   Arv* p = (Arv*)malloc(sizeof(Arv));
   p->info = c;
   p->esq = sae;
   p->dir = sad;
   return p;
}

void imprime(Arv* a) {
   if (!vazia(a)) {
       printf("%d ", a->info); /* mostra raiz */
       imprime(a->esq);        /* mostra sae  */
       imprime(a->dir);        /* mostra sad  */
   }
}

// ==========================================
// 2. RESOLUÇÃO DA PARTE 1: NOVAS FUNÇÕES
// ==========================================

// 1. Função Pertence
int pertence(Arv* a, int x) {
    if (vazia(a)) {
        return 0;
    }
    return a->info == x || pertence(a->esq, x) || pertence(a->dir, x);
}

// 2. Contar Nós
int conta_nos(Arv* a) {
    if (vazia(a)) {
        return 0;
    }
    return 1 + conta_nos(a->esq) + conta_nos(a->dir);
}

// 3. Contar Folhas
int conta_folhas(Arv* a) {
    if (vazia(a)) {
        return 0;
    }
    if (vazia(a->esq) && vazia(a->dir)) {
        return 1;
    }
    return conta_folhas(a->esq) + conta_folhas(a->dir);
}

// ==========================================
// 3. RESOLUÇÃO DA PARTE 2: FUNÇÃO MAIN
// ==========================================

int main() {
    // Montando a árvore de baixo para cima
    Arv* no2 = cria(2, inicia(), inicia());
    Arv* no5 = cria(5, no2, inicia());

    Arv* no30 = cria(30, inicia(), inicia());
    Arv* no20 = cria(20, inicia(), no30);

    Arv* raiz = cria(10, no5, no20);

    // Impressão (Pré-ordem)
    printf("Arvore: ");
    imprime(raiz);
    // Saída esperada: 10 5 2 20 30
    printf("\n");

    // Testando as novas funções
    printf("Pertence o 20? %s\n", pertence(raiz, 20) ? "Sim" : "Nao"); // Sim
    printf("Pertence o 15? %s\n", pertence(raiz, 15) ? "Sim" : "Nao"); // Nao
    printf("Total de nos: %d\n", conta_nos(raiz));       // 5
    printf("Total de folhas: %d\n", conta_folhas(raiz));   // 2 (nós 2 e 30)

    return 0;
}
````

### 4.2 Ampliação do Desafio:  Refatoração de TAD e Implementação de CRUD em Árvore Binária de Busca (BST)



Com base na TAD de Árvore Binária fornecida acima, faça as modificações necessárias na estrutura e implemente o que se pede:

### 1. Alteração da Estrutura do Nó
Modifique o campo `info` da árvore para que ele passe a conter uma estrutura contendo as informações de um aluno, definida por:

```c
struct {
    int matricula;
    float nota;
}
```
Nota: O campo que a árvore deve considerar para a ordenação e critérios de busca/remoção obrigatoriamente será a matricula.

2. Refatoração das Funções da TAD
Adapte todas as funções existentes (cria, insere, busca, retira e as funções de imprime) para que funcionem corretamente com o novo tipo de dado composto do Aluno.

3. Construção de um CRUD
Crie uma interface de menu interativo na função int main() que ofereça ao usuário um CRUD completo para o cadastramento e gerenciamento dos alunos na árvore, contendo:

* [C]reate: Inserir um novo aluno (matrícula e nota) na árvore.
* [R]ead: Buscar um aluno por matrícula (exibindo seus dados) e listar os alunos existentes.
* [U]pdate: Localizar um aluno por matrícula e permitir a alteração de sua nota.
* [D]elete: Remover um aluno da árvore por meio da sua matrícula.

````python
%%writefile desafio.c
#include <stdio.h>
#include <stdlib.h>

// ==========================================
// 1. ESTRUTURAS DE DADOS
// ==========================================

typedef struct {
    int matricula;
    float nota;
} Aluno;

struct arv {
    Aluno info; // Agora armazena os dados do Aluno
    struct arv* esq;
    struct arv* dir;
};
typedef struct arv Arv;

// ==========================================
// 2. FUNÇÕES DA TAD ADAPTADAS
// ==========================================

int vazia(Arv* a) {
    return a == NULL;
}

Arv* inicializa(void) {
    return NULL;
}

// Alterado para receber o Aluno completo
Arv* cria(Aluno al, Arv* esq, Arv* dir) {
    Arv* p = (Arv*)malloc(sizeof(Arv));
    p->info = al;
    p->esq = esq;
    p->dir = dir;
    return p;
}

// Imprime Matrícula e Nota de forma simples
void imprime(Arv* a) {
    if (!vazia(a)) {
        printf("[Mat: %d, Nota: %.1f] ", a->info.matricula, a->info.nota);
        imprime(a->esq);
        imprime(a->dir);
    }
}

// Imprime em formato de árvore hierárquica por matrícula
void imprimec(Arv* a) {
    printf("<");
    if (!vazia(a)) {
        printf("%d ", a->info.matricula);
        imprimec(a->esq);
        printf(",");
        imprimec(a->dir);
    }
    printf(">");
}

void imprimepo(Arv* a) {
    if (!vazia(a)) {
        imprimepo(a->esq);
        imprimepo(a->dir);
        printf("[Mat: %d, Nota: %.1f] ", a->info.matricula, a->info.nota);
    }
}

void imprimeos(Arv* a) {
    if (!vazia(a)) {
        imprimeos(a->esq);
        printf("[Mat: %d, Nota: %.1f] ", a->info.matricula, a->info.nota);
        imprimeos(a->dir);
    }
}

Arv* libera(Arv* a) {
    if (!vazia(a)) {
        libera(a->esq);
        libera(a->dir);
        free(a);
    }
    return NULL;
}

// Busca baseada na matrícula. Retorna o ponteiro do nó se achar, ou NULL.
Arv* busca(Arv* a, int mat) {
    if (vazia(a))
        return NULL;
    if (a->info.matricula == mat)
        return a;

    // Como a árvore mantém propriedades de BST (Busca Binária), otimizamos a busca:
    if (mat < a->info.matricula)
        return busca(a->esq, mat);
    else
        return busca(a->dir, mat);
}

// Inserção baseada na matrícula do Aluno
Arv* insere(Arv* a, Aluno al) {
    if (a == NULL)
        a = cria(al, NULL, NULL);
    else if (al.matricula < a->info.matricula)
        a->esq = insere(a->esq, al);
    else if (al.matricula > a->info.matricula)
        a->dir = insere(a->dir, al);
    else
        printf("\n[Erro] Matricula %d ja cadastrada.\n", al.matricula);
    return a;
}

// Remoção baseada na matrícula
Arv* retira(Arv* r, int mat) {
    if (r == NULL)
        return NULL;

    if (r->info.matricula > mat)
        r->esq = retira(r->esq, mat);
    else if (r->info.matricula < mat)
        r->dir = retira(r->dir, mat);
    else { /* Achou o aluno */
        if (r->esq == NULL && r->dir == NULL) {
            free(r);
            r = NULL;
        }
        else if (r->esq == NULL) {
            Arv* t = r;
            r = r->dir;
            free(t);
        }
        else if (r->dir == NULL) {
            Arv* t = r;
            r = r->esq;
            free(t);
        }
        else { /* Tem dois filhos */
            Arv* pai = r;
            Arv* f = r->esq;
            while (f->dir != NULL) {
                pai = f;
                f = f->dir;
            }
            // Copia a struct Aluno inteira para o nó atual
            r->info = f->info;
            // Configura a matrícula antiga de busca para remover o nó duplicado abaixo
            f->info.matricula = mat;
            r->esq = retira(r->esq, mat);
        }
    }
    return r;
}

// ==========================================
// 3. INTERFACE DO CRUD / MENU
// ==========================================

void menu() {
    printf("\n==== CRUD ALUNOS (ARVORE BINARIA) ====\n");
    printf("1. Cadastrar Aluno (CREATE)\n");
    printf("2. Buscar Aluno (READ)\n");
    printf("3. Atualizar Nota (UPDATE)\n");
    printf("4. Remover Aluno (DELETE)\n");
    printf("5. Listar Alunos (Ordem Simetrica)\n");
    printf("6. Visualizar Estrutura da Arvore\n");
    printf("0. Sair\n");
    printf("Escolha uma opcao: ");
}

int main() {
    Arv* raiz = inicializa();
    int opcao, mat;
    Aluno al;
    Arv* aux;

    do {
        menu();
        scanf("%d", &opcao);

        switch(opcao) {
            case 1: // CREATE
                printf("\n--- Cadastrar Aluno ---\n");
                printf("Digite a matricula: ");
                scanf("%d", &al.matricula);
                printf("Digite a nota: ");
                scanf("%f", &al.nota);
                raiz = insere(raiz, al);
                break;

            case 2: // READ
                printf("\n--- Buscar Aluno ---\n");
                printf("Digite a matricula para busca: ");
                scanf("%d", &mat);
                aux = busca(raiz, mat);
                if (aux != NULL) {
                    printf("Aluno Encontrado! Matricula: %d | Nota: %.1f\n", aux->info.matricula, aux->info.nota);
                } else {
                    printf("Aluno com matricula %d nao encontrado.\n", mat);
                }
                break;

            case 3: // UPDATE
                printf("\n--- Atualizar Nota ---\n");
                printf("Digite a matricula do aluno: ");
                scanf("%d", &mat);
                aux = busca(raiz, mat);
                if (aux != NULL) {
                    printf("Nota atual: %.1f. Digite a nova nota: ", aux->info.nota);
                    scanf("%f", &aux->info.nota);
                    printf("Nota atualizada com sucesso!\n");
                } else {
                    printf("Aluno nao encontrado.\n", mat);
                }
                break;

            case 4: // DELETE
                printf("\n--- Remover Aluno ---\n");
                printf("Digite a matricula para remover: ");
                scanf("%d", &mat);
                aux = busca(raiz, mat);
                if (aux != NULL) {
                    raiz = retira(raiz, mat);
                    printf("Aluno removido com sucesso.\n");
                } else {
                    printf("Aluno nao cadastrado.\n", mat);
                }
                break;

            case 5: // LIST (Read All)
                printf("\n--- Lista de Alunos em Ordem Simetrica ---\n");
                if (vazia(raiz)) printf("Arvore vazia.\n");
                else {
                    imprimeos(raiz);
                    printf("\n");
                }
                break;

            case 6: // Visualização do formato <a <s>, <d>>
                printf("\n--- Estrutura Textual da Arvore ---\n");
                imprimec(raiz);
                printf("\n");
                break;

            case 0:
                printf("\nLiberando memoria e saindo...\n");
                raiz = libera(raiz);
                break;

            default:
                printf("\nOpcao invalida!\n");
        }
    } while(opcao != 0);

    return 0;
}
````
