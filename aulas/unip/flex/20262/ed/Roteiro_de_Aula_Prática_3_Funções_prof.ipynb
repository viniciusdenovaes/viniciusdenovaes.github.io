**Recursividade:** Conceito de uma função que chama a si mesma.
* Elementos essenciais: caso base (condição de parada) e passo recursivo (chamada da função com um subproblema).
* A recursividade é a possibilidade de uma função se chamar, ou, então, voltar a ser chamada após a execução de funções que ela mesma chamou.
* A cada chamada, o programa abre um novo espaço na memória para a sua execução.
* A cada chamada recursiva, é guardada uma cópia dos parâmetros, de modo que não percamos os valores dos parâmetros das chamadas anteriores.
* Em princípio, enquanto houver memória para a recursividade, a função poderá ser reutilizada.
* As funções recursivas têm duas características importantes:
  * ponto de parada: geralmente, um limite superior ou inferior da regra geral;
  * regra geral: reduz a resolução do problema por meio da invocação recursiva de casos menores.

### 3.3. Demonstração Prática (30 minutos)
**Função**
Exemplo: Monte uma função que tem como objetivo receber dois valores inteiros e retornar a soma entre eles.

````c
%%writefile exemplo201.c
#include<stdio.h>
int soma(int a, int b){
  return a+b;
}
int main( ){
  printf("%d \n", soma(3,4));
  printf("%d \n", soma(30,40));
  printf("%d \n", soma(1,9));
return 0;
}
````

````c
%%shell
gcc exemplo201.c -w -o exemplo201
./exemplo201
````

Cada vez que a função `soma` é chamada:
* Os valores informados (por exemplo, 3 e 4) são copiados para os parâmetros `a` e `b`.
* Isso é chamado de passagem por valor.
* A função realiza o cálculo com essas cópias e retorna o resultado.
* O valor retornado é então usado pelo `printf`.

**Função recursiva**
Exemplo: faça uma função recursiva para mostrar os n primeiros elementos de uma PA com o elemento inicial 3, razão 4.

````c
%%writefile exemplo014.c
#include <stdio.h>
// Função recursiva para mostrar os n primeiros termos da PA
void imprimirPA(int atual, int razao, int n) {
  if (n <= 0) {
    return; // caso base: para a recursão
  }
  printf("%d ", atual);
  // chama a função novamente com o próximo termo
  imprimirPA(atual + razao, razao, n - 1);
}
int main() {
  int primeiro=3, razao=4, n=3;
  printf("PA: ");
  imprimirPA(primeiro, razao, n);
  return 0;
}
````

````c
%%shell
gcc exemplo014.c -w -o exercicio201
./exercicio201
````

* O código usa recursão linear (uma chamada por execução).
* Cada chamada: imprime um termo, calcula o próximo.
* O caso base garante que a função não entre em loop infinito.
Esse tipo de solução é elegante para fins didáticos, mas em situações reais uma estrutura de repetição (for/while) costuma ser mais eficiente em termos de memória.

### 4.1 Desafio Inicial: modularizando o fatorial


Considere o algoritmo:

```

algoritmo Fatorial
var
  n,i,fat:inteiro
inicio
  n=5
  fat=1
  para i de 1 ate n faca
    fat=fat*i
  fimpara
  escreva ("fatorial de ",n,"=",fat)
fimalgoritmo
```
 converta em uma função normal fat(n) para calcular o fatorial.




````c
%%writefile desafio.c
#include<stdio.h>
int fat(int n){
  int f=1;
  for(int i=1;i<=n;i++)
    f*=i;
  return f;
}
int main(){
  int a;
  printf("entre com o numero");
  scanf("%d",&a);
  printf("%d!=%d\n",a,fat(a));
  return 0;
}
````

Reescreva a função convertendo numa função recursiva fatR(n)

````c
%%writefile desafio.c
#include<stdio.h>
int fat(int n){
  if(n == 0 || n == 1){
    return 1; // caso base
  }
  return n * fat(n - 1); // chamada recursiva
}
int main(){
  int a;
  printf("entre com o numero");
  scanf("%d",&a);
  printf("%d!=%d\n",a,fat(a));
  return 0;
}
````

## 8. Conclusão

A aula prática sobre **Funções e Recursividade** em C possibilitou aos alunos compreenderem, de forma teórica e aplicada, dois conceitos fundamentais da programação estruturada.
