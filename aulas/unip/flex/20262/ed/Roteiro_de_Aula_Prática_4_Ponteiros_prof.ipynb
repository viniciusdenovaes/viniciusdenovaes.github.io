### Ponteiro de variáveis

   A linguagem C tem uma maneira especial de uma variável armazenar endereços


```
<tipo> *nome;
int *p;

```



  
   O programa reserva um espaço na memória para uma variável chamado _p_ que irá guardar um endereço, e neste endereço armazenado conterá uma informação do tipo inteiro. O símbolo _*_ identifica que uma variável é do tipo ponteiro.



### Para acessar os endereços de memória, a linguagem oferece dois operadores unários

>**_&_** ("endereço de"), aplicado a variáveis, resulta no endereço da posição da memória reservada para a variável

>**_*_** ("conteúdo de"), aplicado a variáveis do tipo ponteiro, acessa o conteúdo do endereço de memória armazenado pela variável ponteiro  

Operador | Operação
---|---
"*" | Declara um ponteiro: int *x
"*" | Conteúdo para onde o ponteiro aponta: int y = *x;
"&" | Endereço onde uma variável está guardada na memória:




###Passagem de valores por valor e por referência



1.   Passagem de valor por valor

A função recebe uma cópia do valor da variável. Isso significa que qualquer alteração dentro da função não afeta a variável original.

2.   Passagem por referência (usando ponteiros)

Em C, a passagem por referência é feita indiretamente, usando ponteiros. A função recebe o endereço da variável, podendo alterar o valor original.



### Alocação de memória

A biblioteca stdlib.h possui algumas funções que nos permitem criar e trabalhar dinamicamente, durante a execução de um certo trecho do programa.


*   malloc()
*   free()
*   sizeof()






o malloc(int), memory allocation ou alocação de memória. A função malloc reserva a quantidade de bytes que é passado como parâmetro e retorna o endereço em que este espaço de memória foi reservado.

```void *malloc (unsigned int num);```

A função malloc() recebe 1 parâmetro de entrada

* num: o tamanho do espaço de memória a ser alocado.

e retorna

* NULL: no caso de erro;
* O ponteiro para a primeira posição do array alocado.
Exemplo reservar 20 bytes na memória:

```malloc(20)```

Note que a função malloc() retorna um ponteiro genérico (void). Esse ponteiro pode ser atribuído a qualquer tipo de ponteiro via type cast.


* Para descobrir quantos bytes um tipo ocupa a biblioteca nos fornece a função sizeof(tipo), tamanho de. Esta função retorna um número inteiro que um tipo ocupa.
```
sizeof nome_da_variável
sizeof (nome_do_tipo)
```



* modificador Cast



```
int *v;
v= (int *)malloc(sizeof(int));
```



* Desalocar, ou liberar, a memória previamente alocada faz com que ela se torne novamente disponível para futuras alocações. Para liberar um bloco de memória previamente alocado utilizamos a função free() cujo protótipo é:


```
void free (void *p);
```


A função free() recebe apenas um parâmetros de entrada: o ponteiro para o início do bloco de memória alocado.

Acesso usando ponteiro
O nome do vetor representa o endereço do primeiro elemento.
Podemos usar aritmética de ponteiros:


```

```
Exemplo equivalente


```
#include <stdio.h>
int main() {
    int v[3] = {10, 20, 30};

    printf("%d\n", *(v + 0));
    printf("%d\n", *(v + 1));
    printf("%d\n", *(v + 2));

    return 0;
}

```



### Ponteiro e Struct

O operador -> em C é usado para acessar membros de uma struct quando você tem um ponteiro para essa struct.

* Se você tem uma struct normal → usa .
* Se você tem um ponteiro para struct → usa ->

* Sem ponteiro (usa .)

```
typedef struct {
    int matricula;
    float nota;
} aluno;

aluno a;
a.matricula = 123;
a.nota = 8.5;
```
Aqui a é uma variável comum, então usamos:
```
a.campo
```

* Com ponteiro (usa ->)


```
aluno *p;
p = (aluno*) malloc(sizeof(aluno));
p->matricula = 123;
p->nota = 8.5;
```

Aqui p é um ponteiro, então usamos:

```
p->campo
```

O operador -> é só uma forma simplificada de escrever:


```
p->nota
```
é exatamente igual a:

```
(*p).nota
```

Situação | Forma correta
---|---
variável struct | a.nota
ponteiro para struct | p->nota
forma equivalente | (*p).nota

### 4.1 Desafio Inicial: Ponteiro e Struct



Crie um programa em C que:

* Aloque dinamicamente memória para um único número inteiro.
* Utilize malloc com sizeof para definir o tamanho correto.
* Use cast para converter o ponteiro retornado por malloc.
* Leia um valor digitado pelo usuário e armazene na memória alocada.
Mostre:

* o valor armazenado
* o endereço de memória

Libere a memória com free.

````python
%%writefile desafio.c
#include <stdio.h>
#include <stdlib.h>

// definicao da struct
typedef struct {
    int matricula;
    float nota;
} aluno;

int main() {
    aluno *p;

    // alocacao dinamica
    p = (aluno *) malloc(sizeof(aluno));

    if (p == NULL) {
        printf("Erro de alocacao!\n");
        return 1;
    }

    // entrada de dados
    printf("Digite a matricula: ");
    scanf("%d", &p->matricula);

    printf("Digite a nota: ");
    scanf("%f", &p->nota);

    // saida
    printf("\nDados do aluno:\n");
    printf("Matricula: %d\n", p->matricula);
    printf("Nota: %.2f\n", p->nota);

    free(p);

    return 0;
}
````

````python
%%shell

gcc desafio.c -w -o programa1

./programa1
````

### 4.2 Ampliação do Desafio: Ponteiros com vetores




Modifique o programa anterior para:

* Ler a quantidade n de alunos de uma turma.
* Alocar dinamicamente um vetor de structs aluno.
* Utilizar malloc com sizeof corretamente.
* Ler os dados de todos os alunos.
* Exibir:
* matrícula e nota de cada aluno
* Liberar a memória.



````python
%%writefile desafio.c
#include <stdio.h>
#include <stdlib.h>

typedef struct {
    int matricula;
    float nota;
} aluno;

int main() {
    aluno *turma;
    int n;

    printf("Digite o numero de alunos: ");
    scanf("%d", &n);

    // alocacao dinamica do vetor
    turma = (aluno *) malloc(n * sizeof(aluno));

    if (turma == NULL) {
        printf("Erro de alocacao!\n");
        return 1;
    }

    // leitura dos dados
    for (int i = 0; i < n; i++) {
        printf("\nAluno %d\n", i + 1);

        printf("Matricula: ");
        scanf("%d", &turma[i].matricula);

        printf("Nota: ");
        scanf("%f", &turma[i].nota);
    }

    // exibicao
    printf("\nLista de alunos:\n");
    for (int i = 0; i < n; i++) {
        printf("Matricula: %d | Nota: %.2f\n",
               turma[i].matricula,
               turma[i].nota);
    }

    free(turma);

    return 0;
}

````

````python
%%shell

gcc desafio.c -w -o programa1

./programa1
````
