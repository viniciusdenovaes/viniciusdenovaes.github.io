## 3. Revisão Conceitual

### 3.1. Introdução ao Google Colab
O Google Colab (ou Colaboratory) é uma plataforma gratuita que permite criar e executar códigos sem a necessidade de instalar programas no computador. Embora seja amplamente utilizado para Python, o ambiente baseado em **Linux** em que ele funciona possibilita o uso de outras linguagens, como **C**.

### 3.2. Processo de Edição (`%%writefile`)
Para criar e salvar arquivos de código fonte nas células, usamos o comando mágico `%%writefile`. Ele deve ser colocado estritamente na **primeira linha da célula**, seguido do nome do arquivo que deseja criar (ex: `%%writefile programa.c`).

### 3.3. Compilação (`%%shell`)
Compilação é o processo onde o compilador (`gcc`) traduz o arquivo de texto para a linguagem nativa do processador Linux, gerando um arquivo executável. O comando mágico `%%shell` transforma a célula inteira do Colab em um terminal Linux completo, permitindo executar comandos em sequência.

### 3.4. Máscaras de Formatação (Tabela de Referência)

| Máscara | Tipo | Descrição |
| :---: | :--- | :--- |
| `%c` | char | Caractere simples |
| `%d` | int | Inteiro decimal |
| `%u` | unsigned int | Inteiro sem sinal |
| `%f` ou `%g` | double ou float | Número real / Ponto flutuante |
| `%e` | double ou float | Notação científica |
| `%s` | char[] / string | Cadeia de caracteres |
| `\n` | - | Quebra de linha |
| `\t` | - | Tabulação |
| `\"` | - | Exibe o caractere com aspas (”) |
| `\\` | - | Exibe o caractere de barra inversa (\) |

## 4. Demonstração Prática

### Passo 1: Criar o programa Hello World e salvar
Execute a célula abaixo para criar o arquivo de código-fonte `teste1.c` no diretório virtual do Colab. *(Dica: clique no ícone da pasta na barra lateral esquerda para ver o arquivo surgir)*.

```python
%%writefile teste1.c
#include <stdio.h>

int main() {
    printf("Ola mundo\n");
    return 0;
}
```

### Passo 2: Compilar e executar o programa
Utilizaremos o `gcc` para compilar.
* A flag `-w` suprime avisos (*warnings*), focando apenas em erros de sintaxe reais.
* A flag `-o output` define o nome do arquivo executável gerado.
* `./output` executa o arquivo gerado.

```python
%%shell
gcc teste1.c -w -o output
./output
```

### Passo 3: Saída com máscaras (`%d` e `%g`)
O `printf` funciona como um "texto com lacunas". O `%g` é uma máscara inteligente para números reais que limpa as casas decimais vazias redundantes.

```python
%%writefile teste1.c
#include <stdio.h> // standard input-output

int main() {
    printf("%d %g\n", 33, 5.3);
    printf("Inteiro = %d    Real = %f\n", 33, 5.3);
    return 0;
}
```

```python
%%shell
gcc teste1.c -w -o output
./output
```

### Passo 4: Leitura Avançada com Máscaras Complexas
O `scanf` pode lidar com formatadores personalizados. No código abaixo, o formato `"%d:%f"` obriga o usuário a digitar o caractere de dois-pontos (`:`) entre os números para que as variáveis sejam preenchidas de forma correta (Ex: `14:30.5`).

```python
%%writefile teste2.c
#include <stdio.h>

int main() {
    int n, h;
    float m;

    printf("Digite um inteiro simples (ex: 10): ");
    scanf("%d", &n);

    printf("Digite um formato de hora (ex: 14:30.5): ");
    scanf("%d:%f", &h, &m);

    printf("Valores lidos: %d e %g\n", n, m);
    printf("Inteiro = %d    Real = %f\n", h, m);
    return 0;
}
```

```python
%%shell
gcc teste2.c -w -o output2
./output2
```

## 5. Atividades Práticas (Desafios)

### 5.1 Desafio Inicial: Realizando a soma
Desenvolva uma aplicação básica que solicite ao usuário a digitação de dois valores e uma operação (`n1 + n2`) e apresente o resultado da soma.

```python
%%writefile desafio.c
#include <stdio.h>

int main() {
    float num1, num2;
    char op;

    printf("Digite a operacao (Ex: 5.2 + 3): \n");
    // O scanf espera: numero [espaco] operador [espaco] numero
    scanf("%f %c %f", &num1, &op, &num2);

    // Exibindo a soma utilizando %g para melhor formatação
    printf("Resultado = %g\n", num1 + num2);
    return 0;
}
```

```python
%%shell
gcc desafio.c -w -o desafio_exec
./desafio_exec
```

### 5.2 Ampliação do Desafio: Calculadora de Quatro Operações
Evolua o desafio inicial transformando-o em uma calculadora completa. O programa deve ler a expressão digitada pelo usuário, identificar o operador e realizar uma das quatro operações aritméticas básicas (`+`, `-`, `*`, `/`) usando a estrutura de controle `switch/case`.

*Atenção: Trate adequadamente o erro de divisão por zero.*

```python
%%writefile calculadora.c
#include <stdio.h>

int main() {
    float num1, num2;
    char op;

    printf("Digite a operacao (Ex: 10.5 * 3): \n");
    scanf("%f %c %f", &num1, &op, &num2);

    switch (op) {
        case '+':
            printf("= %g\n", num1 + num2);
            break;
        case '-':
            printf("= %g\n", num1 - num2);
            break;
        case '*':
            printf("= %g\n", num1 * num2);
            break;
        case '/':
            if (num2 != 0) {
                printf("= %g\n", num1 / num2);
            } else {
                printf("Erro: divisao por zero!\n");
            }
            break;
        default:
            printf("Operador invalido!\n");
            break;
    }
    return 0;
}
```

```python
%%shell
gcc calculadora.c -w -o calc_exec
./calc_exec
```
